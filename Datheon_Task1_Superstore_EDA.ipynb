```python
import pandas as pd

df = pd.read_csv('superstore_sales.csv')

df.head()
```

```python
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())
print("\nMissing values:")
print(df.isnull().sum())
```

```python
df[df['Order Date'].isnull()].head(10)
```

```python
print("Total duplicate rows:", df.duplicated().sum())
```

```python
df[df.duplicated()].head(5)
```

```python
df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)
print("Remaining duplicate rows:", df.duplicated().sum())
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
missing = df.isnull().sum()

print(missing[missing > 0])
```

```python
df[df['Order Date'].isnull()][['Order Date', 'Sales', 'Quantity', 'Discount', 'Profit']].head(10)
```

```python
print("Completely empty rows:", df.isnull().all(axis=1).sum())
```

```python
print("Rows before cleaning:", len(df))

df = df.dropna(subset=['Order Date', 'Sales', 'Quantity', 'Profit'])

print("Rows after cleaning:", len(df))
print("Missing values remaining:", df.isnull().sum().sum())
```

```python
missing = df.isnull().sum()
print(missing[missing > 0])
```

```python
df['Postal Code'] = df['Postal Code'].fillna(-1).astype('int64')

print("Missing values remaining:", df.isnull().sum().sum())
print("Postal Code missing values:", df['Postal Code'].isnull().sum())
```

```python
print("Total Sales:", round(df['Sales'].sum(), 2))
print("Total Profit:", round(df['Profit'].sum(), 2))
print("Total Orders:", df['Order ID'].nunique())
print("Total Products:", df['Product Name'].nunique())
```

```python
category_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)

print(category_sales)
```

```python
import matplotlib.pyplot as plt

category_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)

category_sales.plot(kind='bar', title='Sales by Product Category')

plt.xlabel('Product Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)

region_sales.plot(kind='bar', title='Sales by Region')

plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.show()
```

```python
subcategory_sales = df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False)

subcategory_sales.plot(kind='bar', figsize=(12, 5), title='Sales by Sub-Category')

plt.xlabel('Sub-Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])

monthly_sales = df.groupby(df['Order Date'].dt.to_period('M'))['Sales'].sum()

monthly_sales.plot(kind='line', figsize=(12, 5), title='Monthly Sales Trend')

plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Business Findings
1.Technology has the highest sales among the three product categories.
2.Office Supplies has the lowest sales among the three categories.
3.Sales differ across the four regions.
4.Some product sub-categories generate more sales than others.
5.Monthly sales change over time.
